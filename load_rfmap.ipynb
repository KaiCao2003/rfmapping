```python
fm_unit_ids = set(fm_rfmap_list.unit_ids)
for rfmap in rfmap_list:
    if rfmap.unit_id not in fm_unit_ids:
        print(f"Skipping unit {rfmap.unit_id}: missing from fm_rfmap_list")
        continue

    fm_rfmap = fm_rfmap_list.by_unit_id(rfmap.unit_id)
    newrfmap = rfmap(0.08, 0.16) - rfmap(0.0, 0.08)
    fm_newrfmap = fm_rfmap(0.08, 0.16) - fm_rfmap(0.0, 0.08)

    print(f"Unit {rfmap.unit_id}: session3 RF (80-160 ms minus 0-80 ms)")
    plot_2d_rfmap(newrfmap.to_2d_array())
    print(f"Unit {rfmap.unit_id}: session10 FM RF (80-160 ms minus 0-80 ms)")
    plot_2d_rfmap(fm_newrfmap.to_2d_array())
    print("--------------------")
```
