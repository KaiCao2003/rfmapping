## Before running: choose the saved map

Start Jupyter with the README's `RF_SESSION_DIR`, `RF_DATE`, `RF_SESSION`,
and `RF_PROBES` environment variables set. The file-selection cell uses the
first probe letter and the regular ON-map name for the default `-100_400_1ms`
window. Change `rfmap_path` explicitly to open an OFF/bar file or another
response window. Check the printed full filename before selecting a unit.

Paths must be visible to the computer running this kernel. Native Windows
users should follow the [PowerShell setup](docs/platforms_and_data_safety.md).
Keep filename and directory capitalization exact.

These cells read a saved `.rfmap`, sum time bins, and display an RF; they do
not generate or overwrite the source map. If you add
`rf_2d(..., result_path=...)`, use a separate derived `.npz` cache path.
Changed inputs/settings can replace that cache. Omit `result_path` to keep
detection results in memory, or choose a new path to preserve a previous
result. Never use the source `.rfmap` as a cache destination.

Read the [raw-to-RF guide](README.md) for generating the file. Timing preview
results must be reviewed against the photodiode trace before RF generation;
a plot in this notebook does not replace that critical timing review.


```python
%matplotlib inline
%load_ext autoreload
%autoreload 2

import os
from pathlib import Path
from Utils.rfmap import load_rf_maps, plot_2d_rfmap
```

```python
session_dir = Path(os.environ["RF_SESSION_DIR"])
date = os.environ["RF_DATE"]
session = os.environ["RF_SESSION"]
probe = os.environ["RF_PROBES"][0]

rfmap_dir = session_dir / "data/rfmapping/good/-100_400_1ms" / f"Probe{probe}"
rfmap_path = rfmap_dir / f"regular_unitsSpikeCounts_{date}_{session}.rfmap"
print(rfmap_path)

rfmap_list = load_rf_maps(rfmap_path)
print("Available unit IDs:", rfmap_list.unit_ids)
```

```python
# Replace this with an ID from the printed list to inspect another unit.
unitID = rfmap_list.unit_ids[0]
rfmap = rfmap_list.by_unit_id(unitID)
newrfmap = rfmap(0.08, 0.16) - rfmap(0.0, 0.08)
```

```python
print(f"Unit {unitID}: RF (80–160 ms minus 0–80 ms)")
plot_2d_rfmap(newrfmap.to_2d_array())
```
