# Compare two RF sessions

Set both RF source paths and the unit ID in the configuration cells.
The calculations convert counts to rates before comparing response windows.
For the portable single-map workflow, use `load_rfmap.ipynb`.


```python
%matplotlib inline
%load_ext autoreload
%autoreload 2

from pathlib import Path
from Utils.rflocate import load_rfmap
from Utils.rflocate.plotting import plot_2d_rfmap
```

```python
rfmap_dir = Path("/mnt/senzailab/Kai/#Recording/m19/260821/260827_2/data/rfmapping/good/-100_400_1ms/ProbeA")
rfmap_path = next(rfmap_dir.glob("*.rfmap"))

rfmap_list = load_rfmap(rfmap_path).to_firing_rate(reconstruct_presentations=True)

fm_rfmap_dir = Path("/mnt/senzailab/Kai/#Recording/m19/260827/260827_10/data/rfmapping/good/-100_400_1ms/ProbeA")
fm_rfmap_path = next(fm_rfmap_dir.glob("*.rfmap"))

fm_rfmap_list = load_rfmap(fm_rfmap_path).to_firing_rate(reconstruct_presentations=True)
```

```python
fm_rfmap_path
```

```python
unitID = 243
rfmap = rfmap_list[f"{unitID}"]
newrfmap = rfmap.mean_rate(0.08, 0.16) - rfmap.mean_rate(0.0, 0.08)

fm_rfmap = fm_rfmap_list[f"{unitID}"]
fm_newrfmap = fm_rfmap.mean_rate(0.08, 0.16) - fm_rfmap.mean_rate(0.0, 0.08)
```

```python
print(f"Unit {unitID}: session3 RF (80-160 ms minus 0-80 ms)")
plot_2d_rfmap(newrfmap.to_2d_array())

print(f"Unit {unitID}: session10 FM RF (80-160 ms")
plot_2d_rfmap(fm_rfmap.mean_rate(0.08, 0.16).to_2d_array())

print(f"Unit {unitID}: session10 FM RF (80-160 ms minus 0-80 ms)")
plot_2d_rfmap(fm_newrfmap.to_2d_array())
```

```python
fm_unit_ids = set(fm_rfmap_list.unit_ids)
for rfmap in rfmap_list:
    if rfmap.unit_id not in fm_unit_ids:
        print(f"Skipping unit {rfmap.unit_id}: missing from fm_rfmap_list")
        continue

    fm_rfmap = fm_rfmap_list.by_unit_id(rfmap.unit_id)
    newrfmap = rfmap.mean_rate(0.08, 0.16) - rfmap.mean_rate(0.0, 0.08)
    fm_newrfmap = fm_rfmap.mean_rate(0.08, 0.16) - fm_rfmap.mean_rate(0.0, 0.08)

    print(f"Unit {rfmap.unit_id}: session3 RF (80-160 ms minus 0-80 ms)")
    plot_2d_rfmap(newrfmap.to_2d_array())
    print(f"Unit {rfmap.unit_id}: session10 FM RF (80-160 ms minus 0-80 ms)")
    plot_2d_rfmap(fm_newrfmap.to_2d_array())
    print("--------------------")
```
